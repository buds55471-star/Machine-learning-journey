```python
percentile25 = df['Annual Revenue ($M)'].quantile(0.25)
percentile75 = df['Annual Revenue ($M)'].quantile(0.75)
```

```python
iqr = percentile75 - percentile25
iqr
```

```python
upper_limit = percentile75 + 1.5 * iqr
lower_limit = percentile25 - 1.5 * iqr
print("Upper limit: ", upper_limit)
print("Lower limit: ", lower_limit)
```

```python
#Finding Outliers
outliers = df[(df['Annual Revenue ($M)'] > upper_limit) | (df['Annual Revenue ($M)'] < lower_limit)]
outliers
```

```python
#Trimming
new_df = df[df['Annual Revenue ($M)'] < upper_limit]
```

```python
new_df.shape
```

```python
# Remove outliers permanently from the main DataFrame
df = df[(df['Annual Revenue ($M)'] >= lower_limit) & (df['Annual Revenue ($M)'] <= upper_limit)]

# Visualize the clean distribution
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(df['Annual Revenue ($M)'], kde=True)
plt.title('Cleaned Distribution (No Outliers)')

plt.subplot(1, 2, 2)
sns.boxplot(x=df['Annual Revenue ($M)'])
plt.title('Cleaned Boxplot')
plt.show()

print(f"New dataset shape: {df.shape}")
```
