```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('employee_promotion_prediction.csv')
df.head()
```

```python
df.shape
```

```python
df['education_level'].value_counts()
```

```python
df_encoded = df.copy()
education_encoded = {'Bachelor':0, 'Master':1, 'PhD':2}

df_encoded['Education_Enc'] = df_encoded["education_level"].map(education_encoded)
df_encoded.head()
```

```python
df_encoded = pd.get_dummies(df_encoded, columns=['city_tier'])
df_encoded
```

```python
job_title_counts = df_encoded['department'].value_counts().to_dict()
df_encoded['job_title_counts'] = df_encoded['department'].map(job_title_counts)
df_encoded.head()
```

```python
#Feature scaling
from sklearn.preprocessing import StandardScaler, MinMaxScaler

df_scaled = df_encoded.copy()

standard_scaler = StandardScaler()
df_scaled['age'] = standard_scaler.fit_transform(df_scaled[['age']])

minmax_scaler = MinMaxScaler()
df_scaled['job_title_counts'] = minmax_scaler.fit_transform(df_scaled[['job_title_counts']])

df_scaled.head()
```
