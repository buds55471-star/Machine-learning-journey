### Handling Missing Categorical Values

Given the presence of missing values in 'gender', 'enrolled_university', and 'education_level', we will impute them using the mode (most frequent value) of each column. This is a common strategy for handling missing values in categorical data.

```python
# Impute missing 'gender' values with the mode
mode_gender_train = x_train['gender'].mode()[0]
x_train['gender'] = x_train['gender'].fillna(mode_gender_train)
x_test['gender'] = x_test['gender'].fillna(mode_gender_train) # Use train mode for test set

# Impute missing 'enrolled_university' values with the mode
mode_enrolled_train = x_train['enrolled_university'].mode()[0]
x_train['enrolled_university'] = x_train['enrolled_university'].fillna(mode_enrolled_train)
x_test['enrolled_university'] = x_test['enrolled_university'].fillna(mode_enrolled_train) # Use train mode for test set

# Impute missing 'education_level' values with the mode
mode_education_train = x_train['education_level'].mode()[0]
x_train['education_level'] = x_train['education_level'].fillna(mode_education_train)
x_test['education_level'] = x_test['education_level'].fillna(mode_education_train) # Use train mode for test set

print("Missing values in x_train after imputation:")
print(x_train.isnull().sum())

print("\nMissing values in x_test after imputation:")
print(x_test.isnull().sum())
```

```python
oe = OrdinalEncoder(categories=[['No relevent experience','Has relevent experience'], ['Primary School', 'High School', 'Graduate', 'Masters', 'Phd']])
x_train_ordinal_features = oe.fit_transform(x_train[['relevent_experience', 'education_level']])
x_test_ordinal_features = oe.transform(x_test[['relevent_experience', 'education_level']])
```
